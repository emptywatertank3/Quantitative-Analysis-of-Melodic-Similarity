# Calculate Weighting Scores for Each Element

Our approach to evaluating the distinctive features of a melody involves two methods drawn from the fields of psychology and NLP:

1. **Modified Tversky Measure for Assessing Word Salience**: This measure evaluates the prominence or distinctiveness of each word-like token within the melody.
  
2. **TF-IDF for Evaluating Word Importance and Uniqueness**: This approach captures the significance and rarity of each token, highlighting how unique each element is in the context of the melody.

These weighting scores provide a quantitative basis for assessing the unique and shared features within melodies, contributing to a comprehensive similarity analysis.


## Tversky for Salience

Inspired by the Tversky measure, this study introduces a modified Tversky approach, which incorporates each element’s prevalence and distribution across two sets to assess individual contributions to overall similarity.

Below is the code, and for a more detailed explanation, refer to this: [Modified Tversky Measure for Elements.ipynb](https://colab.research.google.com/drive/1jnwict4GttXxo4metpaXbrTFMxY6yqci?usp=sharing).

```python
# @title Code for modifed-Tversky for element
import numpy as np

def get_tversky(A, B):
    # Find the intersection of elements between sequences A and B (shared elements)
    shared_elements = set(A).intersection(B)
    # Initialize an empty list to store the salience scores for each element in A
    salience_scores = []

    # Lengths of sequences A and B for coefficient calculations
    l_A = len(A)
    l_B = len(B)

    # Coefficients α and β based on sequence lengths
    alpha = l_A / (l_A + l_B)
    beta = l_B / (l_A + l_B)

    # Iterate over each element x in sequence A to calculate its salience score
    for x in A:
        if x in shared_elements:
            # Counts of element x in sequences A and B
            a_x = A.count(x)  # Frequency of x in sequence A
            b_x = B.count(x)  # Frequency of x in sequence B

            # Calculate the numerator (top part of the Tversky measure formula)
            top = a_x / l_A

            # Calculate the denominator (bottom part of the Tversky measure formula)
            bottom = top + alpha * (1 - a_x / l_A) + beta * (1 - b_x / l_B)

            # Calculate the salience score (Tversky measure) for element x
            w = top / bottom
            # Handle cases where w is NaN (this happens if top and bottom are both zero)
            if np.isnan(w):
                w = 0
            # Append the salience score to the list
            salience_scores.append(w)
        else:
            # If x is not in shared elements, its salience score is 0
            salience_scores.append(0)

    # Return the list of salience scores for all elements in sequence A
    return salience_scores


```

```python
# @title Case Example
# Calculate modified Tversky scores for each token in Melody A relative to Melody B
tversky_scores_a = get_tversky(M2Wtoken_a, M2Wtoken_b)

# Calculate modified Tversky scores for each token in Melody B relative to Melody A
tversky_scores_b = get_tversky(M2Wtoken_b, M2Wtoken_a)

# Display the results with formatted Tversky scores rounded to three decimal places
print("Tokenized Melody A:", M2Wtoken_a)
print("Tokenized Melody B:", M2Wtoken_b)
print("\nModified Tversky Scores for Melody A relative to Melody B:", [f"{score:.2f}" for score in tversky_scores_a])
print("Modified Tversky Scores for Melody B relative to Melody A:", [f"{score:.2f}" for score in tversky_scores_b])
```

## TF-IDF for Importance and Uniqueness

TF-IDF (Term Frequency-Inverse Document Frequency) is a well-known measure widely used in NLP to evaluate the importance and uniqueness of words. In this context, we use TF-IDF to highlight significant and rare elements in melodies, quantifying their distinctiveness.

Let’s use the sklearn library to apply TF-IDF to our tokenized melodies.

```python
# @title Code for TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

def get_tfidf_scores(tokens_a, tokens_b):
    """
    Calculate TF and IDF scores for two tokenized melodies.

    Parameters:
    tokens_a (list): Tokenized melody A.
    tokens_b (list): Tokenized melody B.

    Returns:
    tuple: Arrays containing TF and IDF scores for each melody.
    """
    # Convert token lists to single strings for TF-IDF calculation
    tokenized_melody_a = ' '.join(tokens_a)
    tokenized_melody_b = ' '.join(tokens_b)

    # Create a list of tokenized melodies for TF-IDF
    # NOTE: Expand the list of tokenized melodies for better TF-IDF results
    melodies = [tokenized_melody_a, tokenized_melody_b]

    # Initialize the TF-IDF Vectorizer with `use_idf=True` to get both TF and IDF
    vectorizer = TfidfVectorizer(use_idf=True)

    # Fit and transform the data to get TF-IDF scores
    tfidf_matrix = vectorizer.fit_transform(melodies)

    # Extract the term frequencies (TF) for each melody
    tf_a = tfidf_matrix.toarray()[0]  # TF values for each token in melody A
    tf_b = tfidf_matrix.toarray()[1]  # TF values for each token in melody B

    # Extract the IDF scores and apply them to each melody
    idf_scores = vectorizer.idf_
    idf_a = [idf_scores[i] * tf_a[i] for i in range(len(idf_scores))]
    idf_b = [idf_scores[i] * tf_b[i] for i in range(len(idf_scores))]

    # Display TF and IDF scores
    print("TF Scores for Melody A:", [f"{score:.2f}" for score in tf_a])
    print("TF Scores for Melody B:", [f"{score:.2f}" for score in tf_b])
    print("IDF Scores for Melody A:", [f"{score:.2f}" for score in idf_a])
    print("IDF Scores for Melody B:", [f"{score:.2f}" for score in idf_b])


    print("=====================================================================================================")
    print("For more accurate TF-IDF vectorization, consider expanding the limited list of tokenized melodies provided.")
    return tf_a, tf_b, idf_a, idf_b

```

# Calculate Similarity Using Word2Vec Model

To calculate the similarity between two tokenized melodies using a Word2Vec model with weights, follow these steps:

Load the Word2Vec Model: Load the pre-trained model to provide vector representations for each token in the melodies.
Compute Weighted Mean Vectors: For each melody, retrieve the Word2Vec vectors for all tokens and calculate a weighted mean vector. The weights (e.g., term frequencies or importance scores) adjust the contribution of each vector in the mean calculation.
Calculate Cosine Similarity: Using the weighted mean vectors for both melodies, compute cosine similarity to measure how closely the two melodies are related.

```python
# @title Code for Similarity Measure
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def weighted_mean_vector(tokens, weights, model):
    """
    Calculate the weighted mean vector for a sequence of tokens using specified weights.
    Weights are only applied during the mean calculation.
    """
    # Collect vectors for tokens that exist in the model's vocabulary
    vectors = [model.wv[token] for token in tokens if token in model.wv]
    valid_weights = [weights[i] for i, token in enumerate(tokens) if token in model.wv]

    # Calculate the weighted mean vector, applying weights during the mean operation
    if valid_weights:  # Only calculate if there are valid weights
        weighted_mean = np.average(vectors, axis=0, weights=valid_weights)
    else:
        weighted_mean = np.zeros(model.vector_size)  # Return a zero vector if no valid weights

    return weighted_mean

def calculate_weighted_similarity(tokens_a, tokens_b, weights_a, weights_b, model):
    """
    Calculate the cosine similarity between two weighted mean vectors.
    """
    # Calculate weighted mean vector for each melody
    mean_vector_a = weighted_mean_vector(tokens_a, weights_a, model)
    mean_vector_b = weighted_mean_vector(tokens_b, weights_b, model)

    # Calculate and return cosine similarity
    similarity = cosine_similarity([mean_vector_a], [mean_vector_b])[0][0]
    return similarity

def calculate_custom_weights(tf_scores, idf_scores=None, tversky_scores=None):
    """
    Calculate custom weights by multiplying TF, IDF, and Tversky scores after adding one to each.
    Each weight has 1 added to avoid zero multiplication.

    Parameters:
    tf_scores (list): Term Frequency scores.
    idf_scores (list or None): Inverse Document Frequency scores. Set to None to exclude.
    tversky_scores (list or None): Tversky scores. Set to None to exclude.

    Returns:
    list: Calculated custom weights.
    """
    max_length = len(tf_scores)

    # Ensure idf_scores and tversky_scores are lists of the correct length
    if idf_scores is None:
        idf_scores = [0] * max_length
    else:
        idf_scores = idf_scores + [0] * (max_length - len(idf_scores))

    if tversky_scores is None:
        tversky_scores = [0] * max_length
    else:
        tversky_scores = tversky_scores + [0] * (max_length - len(tversky_scores))

    # Calculate custom weights with 1 added to each term to prevent zero multiplication
    custom_weights = [(1 + tf) * (1 + idf) * (1 + tv)
                      for tf, idf, tv in zip(tf_scores, idf_scores, tversky_scores)]

    # Normalize custom weights
    total_weight = sum(custom_weights)
    if total_weight > 0:
        norm_custom_weights = [w / total_weight for w in custom_weights]
    else:
        norm_custom_weights = [0] * max_length  # All weights are zero

    return norm_custom_weights
```
